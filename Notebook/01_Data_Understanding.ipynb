```python
import pandas as pd
import numpy as np
```

```python
# Load the dataset

df = pd.read_csv("../Data/Raw/ChargingRecords.csv")
```

```python
# Checking the dataset dimensions

print("Row :" ,df.shape[0])
print("Columns :" ,df.shape[1])
```

```python
# Show the the records

df.head()
```

```python
# Dataset info

df.info()
```

```python
# Finding any missing values

print(df.isnull().sum())
```

```python
# Duplicate values

print("Duplicate values: " , df.duplicated().sum())
```

```python
# Statistical summary
display(df.describe())
```

```python
# Displaying the duplicate values

duplicate_values = df[df.duplicated(keep=False)]

display(duplicate_values.sort_values(by=df.columns.to_list()))
```

```python
# Finding any negative duration

negative_duration =(df["Duration"] < 0).sum()

display(negative_duration)
```

```python
negative_records = df[df["Duration"]<0]

display(
    negative_records[
        ["StartDatetime", "EndDatetime", "Duration", "StartTime", "EndTime"]
    ].head(20)
    )
```

```python
# Find the 0 durations records

zero_duration = (df["Duration"] == 0).sum()

display(zero_duration)
```

```python
print(df["Demand"].describe())

# Check negative demand
print("Negative Demand:", (df["Demand"] < 0).sum())

# Check zero demand
print("Zero Demand:", (df["Demand"] == 0).sum())

# Check unique demand values
print("Unique Demand Values:", df["Demand"].nunique())
```

```python
categorical_columns = [
    "Location",
    "ChargerType",
    "ChargerCompany"
]

for column in categorical_columns:
    print(f"\n{column}")
    print("Unique values:", df[column].nunique())
    print(df[column].value_counts())
```
